```python
#preprocessing
from sklearn.preprocessing import LabelEncoder
```

```python
le=LabelEncoder()
```

```python
for i in df.columns:
  if df[i].dtype=='object':
    df[i]=le.fit_transform(df[i])
```

```python
df.head()
```

```python
X = df.drop('y', axis=1)
y = df['y']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=1, stratify=y)
```

```python
#undersampling on the training data
```

# Model Selection

# Modeling and Evaluation

```python
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
```

```python
rf = RandomForestClassifier()
```

```python
param = {'n_estimators': [50,100,120], 'bootstrap': [True, False]}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
gsc = GridSearchCV(estimator=rf, param_grid=param, scoring = 'accuracy')
```

```python
gsc.fit(X_train, y_train)
```

```python
y_pred = gsc.best_estimator_.predict(X_test)
```

```python
from sklearn.metrics import classification_report
```

```python
print(classification_report(y_test, y_pred))
```

# XGBoost

```python
param = { 'learning_rate': [0.15,0.20,0.25,0.30],
          'max_depth': [3,4,5],
          'min_child_weight':[1,3,5,7],
          'n_estimators': [50,100,120]
          }
```

```python
xgb = XGBClassifier()
```

```python
gsc = GridSearchCV(estimator=xgb, param_grid=param, scoring = 'accuracy')
```

```python
gsc.fit(X_train, y_train)
```

```python
y_pred = gsc.best_estimator_.predict(X_test)
```

```python
print(classification_report(y_test, y_pred)) #xgboost
```
